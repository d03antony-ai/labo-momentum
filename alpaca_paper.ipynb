```python
!pip install -q alpaca-py

from google.colab import userdata
from alpaca.trading.client import TradingClient

client = TradingClient(
    userdata.get("APCA_API_KEY_ID"),
    userdata.get("APCA_API_SECRET_KEY"),
    paper=True,
)

compte = client.get_account()
print("Statut :", compte.status)
print("Cash :", compte.cash)
```

```python
TRADING_ACTIF = False
print("Trading actif :", TRADING_ACTIF)
```

```python
from alpaca.trading.requests import MarketOrderRequest
from alpaca.trading.enums import OrderSide, TimeInForce

if TRADING_ACTIF:
    ordre = client.submit_order(
        MarketOrderRequest(
            symbol="SPY",
            qty=1,
            side=OrderSide.BUY,
            time_in_force=TimeInForce.DAY,
        )
    )
    print("Ordre envoyé :", ordre.symbol, ordre.qty, ordre.status)
else:
    print("Trading désactivé, aucun ordre envoyé.")
```

```python
for p in client.get_all_positions():
    print(p.symbol, p.qty, "parts, prix moyen", p.avg_entry_price, "$, P&L", p.unrealized_pl, "$")
```

```python
if TRADING_ACTIF:
    client.close_position("SPY")
    print("Position SPY vendue.")
else:
    print("Trading désactivé, aucune vente.")
```

```python
compte = client.get_account()
print("Cash :", compte.cash)
print("Valeur totale :", compte.equity)
```

```python
from alpaca.trading.requests import GetOrdersRequest
from alpaca.trading.enums import QueryOrderStatus
filtre = GetOrdersRequest(status=QueryOrderStatus.ALL, limit=10)
ordres = client.get_orders(filter=filtre)

for o in ordres:
    print(o.symbol, o.side, o.qty, o.status, o.filled_avg_price)
```

```python
client.cancel_orders()
```
